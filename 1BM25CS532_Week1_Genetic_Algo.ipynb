```python
import random


def fitness(x):
    return x**2


pop = [random.randint(0, 31) for _ in range(20)]

for _ in range(30):
    scores = [fitness(x) for x in pop]
    parents = random.choices(pop, weights=scores, k=20)
    next_gen = []

    for i in range(0, 20, 2):
        p1, p2 = parents[i], parents[i + 1]

        if random.random() < 0.8:
            pt = random.randint(1, 4)
            c1 = ((p1 >> pt) << pt) | (p2 & ((1 << pt) - 1))
            c2 = ((p2 >> pt) << pt) | (p1 & ((1 << pt) - 1))
        else:
            c1, c2 = p1, p2

        if random.random() < 0.1:
            c1 ^= 1 << random.randint(0, 4)
        if random.random() < 0.1:
            c2 ^= 1 << random.randint(0, 4)

        next_gen.extend([c1 % 32, c2 % 32])

    pop = next_gen

best = max(pop, key=fitness)
print(best, fitness(best))
```
