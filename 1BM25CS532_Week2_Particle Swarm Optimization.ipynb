```python
import random


def fuel_cost(pos):
    return pos[0] ** 2 + pos[1] ** 2


particles = [
    [random.uniform(-10, 10), random.uniform(-10, 10)] for _ in range(15)
]
velocities = [
    [random.uniform(-1, 1), random.uniform(-1, 1)] for _ in range(15)
]

pbest_pos = [p[:] for p in particles]
pbest_val = [fuel_cost(p) for p in particles]

gbest_pos = min(pbest_pos, key=fuel_cost)
gbest_val = fuel_cost(gbest_pos)

w, c1, c2 = 0.5, 1.5, 1.5

for _ in range(30):
    for i in range(15):
        for d in range(2):
            r1, r2 = random.random(), random.random()
            velocities[i][d] = (
                w * velocities[i][d]
                + c1 * r1 * (pbest_pos[i][d] - particles[i][d])
                + c2 * r2 * (gbest_pos[d] - particles[i][d])
            )
            particles[i][d] += velocities[i][d]

        cost = fuel_cost(particles[i])
        if cost < pbest_val[i]:
            pbest_val[i], pbest_pos[i] = cost, particles[i][:]
            if cost < gbest_val:
                gbest_val, gbest_pos = cost, particles[i][:]

print(gbest_pos, gbest_val)
```
