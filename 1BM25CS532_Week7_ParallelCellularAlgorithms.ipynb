```python
from concurrent.futures import ThreadPoolExecutor

def denoise_pixel(args):
    x, y, image = args
    total = 0
    for i in range(-1, 2):
        for j in range(-1, 2):
            total += image[x + i][y + j]
    return x, y, round(total / 9)

def parallel_cellular_denoise(image):
    width = len(image)
    height = len(image[0])
    new_image = [row[:] for row in image]
    
    tasks = [
        (x, y, image)
        for x in range(1, width - 1)
        for y in range(1, height - 1)
    ]
    
    with ThreadPoolExecutor() as executor:
        for x, y, new_val in executor.map(denoise_pixel, tasks):
            new_image[x][y] = new_val
            
    return new_image

if __name__ == "__main__":
    noisy_image = [
        [50, 52, 48, 51, 49],
        [49, 50, 51, 50, 52],
        [51, 48, 250, 49, 50],
        [50, 52, 49, 51, 48],
        [48, 50, 51, 49, 50]
    ]

    cleaned_image = parallel_cellular_denoise(noisy_image)

    for row in cleaned_image:
        print(row)
```
