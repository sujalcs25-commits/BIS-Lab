```python
import random


def grid_cost(x, y):
    return (x - 5) ** 2 + (y - 3) ** 2


def express_gene(dna):
    return [dna[0] * 0.5, dna[1] * 0.5]


pop = [[random.randint(0, 20), random.randint(0, 20)] for _ in range(20)]

for _ in range(50):
    scores = [1 / (1 + grid_cost(*express_gene(dna))) for dna in pop]
    parents = random.choices(pop, weights=scores, k=20)
    next_gen = []

    for i in range(0, 20, 2):
        p1, p2 = parents[i], parents[i + 1]
        c1 = [p1[0], p2[1]] if random.random() < 0.8 else p1[:]
        c2 = [p2[0], p1[1]] if random.random() < 0.8 else p2[:]

        if random.random() < 0.1:
            c1[random.randint(0, 1)] += random.choice([-1, 1])
        if random.random() < 0.1:
            c2[random.randint(0, 1)] += random.choice([-1, 1])

        next_gen.extend([c1, c2])

    pop = next_gen

best_dna = min(pop, key=lambda d: grid_cost(*express_gene(d)))
best_strategy = express_gene(best_dna)

print(f"Optimal Grid Load Controls: x = {best_strategy[0]}, y = {best_strategy[1]}")
print(f"Minimum Grid Imbalance/Cost: {grid_cost(*best_strategy)}")
```
